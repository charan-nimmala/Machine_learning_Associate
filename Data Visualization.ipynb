### Line Plot

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 16]

plt.plot(x, y, marker='+', linestyle='dotted', color='black') 
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# plt.grid(True)  
plt.show()
```

```python

x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 18, 16]

fig,axis=plt.subplots()
fig.set_facecolor('green')
axis.set_facecolor('blue')


axis.plot(x, y, marker='o', linestyle='-', color='r') 
axis.set_title("Line Plot")
axis.set_xlabel("X-axis")
axis.set_ylabel("Y-axis")
  
plt.show()
```

### subplot:
- Adds a subplot (i.e., a plot panel) inside the figure.
- Syntax: plt.subplot(nrows, ncols, index)

```python
plt.subplot(1, 2, 1)  # 1 row, 2 columns, 1st subplot
plt.plot([1, 2, 3], [4, 5, 6])

plt.subplot( )  # 1 row, 2 columns, 2nd subplot
plt.plot([1, 2, 3], [6, 5, 4])

plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 4))

# First subplot
plt.subplot(1, 2, 1)
plt.plot([1, 2, 3], [4, 5, 6])
plt.title("Plot 1")

# Second subplot
plt.subplot(1, 2, 2)
plt.plot([1, 2, 3], [6, 5, 4])
plt.title("Plot 2")
plt.show() 
```

## Advanced Matplotlib

### Histogram
- Shows the distribution of a single numeric variable.
- Groups values into bins and counts the number of values in each bin.

```python
import matplotlib.pyplot as plt
import numpy as np

data = np.random.randn(1000)

plt.hist(data, bins=30, color='green', edgecolor='black')
plt.title("Histogram")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

### Boxplot
- Summarizes distribution using 5 statistics: min, Q1, median, Q3, max.
- Useful for identifying outliers.

```python
import seaborn as sns

sns.boxplot(data=data, orient='v', color='orange')   # 'h' - Horizontal Boxplot, 'v' - Verticle Boxplot.
plt.title("Boxplot") 
plt.show()
```

### savefig, DPI, formats

```python
import matplotlib.pyplot as plt
import numpy as np

# Creating some sample data
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.savefig('plot.png', dpi=300)
plt.savefig('plot.pdf', dpi=150)
plt.savefig('plot.jpg', dpi=200)
plt.savefig('plot.svg')  # Scalable Vector Graphics
plt.savefig('plot.eps')  # Encapsulated PostScript
plt.show()
```

```python
# Save the plot with different options
# Save as a PNG file with 300 DPI (dots per inch)
plt.savefig('plot.png', dpi=300)
plt.savefig('plot.pdf', dpi=150)
plt.savefig('plot.jpg', dpi=200)

# Save as SVG file (vector format, no need for DPI)
plt.savefig('plot.svg')

# Save as EPS (Encapsulated PostScript, for high-quality printing)
plt.savefig('plot.eps')

# To save in a different format, we can specify the extension in the file name (e.g., .png, .pdf, .jpg, .svg, .eps).
```

```python
# Save the plot with a tight layout (removes extra space around the figure)
plt.savefig('plot.png', dpi=300, bbox_inches='tight')
```

## Seaborn Themes & Styles

### set_theme()
- The set_theme() function allows us to customize the appearance of our seaborn plots with predefined settings.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set the overall theme for the plots
sns.set_theme(style="white", palette="deep", font="Arial", rc={"figure.figsize": (8, 6)})
```

```python
# Example plot to visualize the theme
tips = sns.load_dataset("tips")
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

#### Common Parameters of set_theme():
**style:** Controls the background style of the plot.
- Options: 'white', 'dark', 'whitegrid', 'darkgrid'.

**palette:** Defines the color palette used for the plot.
- we can choose from built-in palettes like 'deep', 'muted', 'pastel', etc., or specify a custom palette.

  **Built-in Palettes:** 
  - deep: Default palette with 6 colors.
  - muted: A set of soft colors.
  - pastel: Lighter colors.
  - dark: More saturated colors.
  - colorblind: Palette designed for colorblind-friendly plots.

**font:** Sets the font for the plot.
- Example: "Arial", "Comic Sans MS", "Times New Roman".

**rc:** Allows you to set custom properties like figure size, axis label size, etc.
- Example: {"figure.figsize": (8, 6)} to set the size of the figure.

### Seaborn Statistical Plots

- Used to understand distribution, relationships, and correlations.

```python
df = sns.load_dataset("titanic")  
df.head()
```

#### Histogram / Distribution Plot

```python
sns.histplot(df["age"], kde=True)
plt.title("Age Distribution")
plt.show()
```

#### KDE Plot (smooth density curve)

```python
sns.kdeplot(df["fare"])
plt.title("Fare Density")
plt.show()
```

#### Scatter Plot (two numeric variables)

```python
sns.scatterplot(x="age", y="fare", data=df)
plt.title("Age vs Fare")
plt.show()
```

#### Regression Plot (trend line)

```python
sns.regplot(x="age", y="fare", data=df)
plt.title("Regression: Fare vs Age")
plt.show()
```

### Seaborn Categorical Plots
- Used to compare categories like Gender, Class, Survival.

### Count Plot

```python
sns.countplot(x="sex", data=df)
plt.title("Male vs Female Count")
plt.show()
```

### Bar Plot – shows mean of a numeric column per category

```python
sns.barplot(x="class", y="fare", data=df)
plt.title("Average Fare by Class")
plt.show()
```

### Box Plot – spread of values (min, Q1, median, Q3, max)

```python
sns.boxplot(x="class", y="age", data=df)
```

### Violin Plot – distribution + density

```python
sns.violinplot(x="sex", y="age", data=df)
```

## Seaborn Multi-Plot Grids
- Used for comparative analysis across multiple distributions/relationships.

### Pair Plot (relationship matrix)

```python
sns.pairplot(df)
plt.show()
```

### FacetGrid (subplots by category)

```python
g = sns.FacetGrid(df, col="sex")
g.map(sns.histplot, "age", kde=True)
plt.show()
```

### Joint Plot (scatter + distributions)

```python
sns.jointplot(x="age", y="fare", data=df, kind="hex")
plt.show()
```

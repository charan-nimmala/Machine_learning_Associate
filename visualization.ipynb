# Introduction to Matplotlib

**Matplotlib** is a popular plotting library in Python. It was originally created by **John D. Hunter** in **2003** as a patch to IPython for enabling interactive MATLAB-style plotting via Gnuplot from the IPython command line. Python was still in its early days, and Matplotlib was one of the first libraries to offer robust capabilities for creating a wide range of static, animated, and interactive visualizations in Python. It is particularly useful for those working in Python for data visualization and scientific computing. Over the years, Matplotlib has evolved into a powerful library capable of producing complex plots with fine-tuned details.

The library is designed to be understandable for beginners but powerful enough for experts. Matplotlib allows for the creation of **bar charts, scatter plots, histograms, error charts, power spectra, and much more**, with just a few lines of code.

# 📊 Graphs and Plots using Matplotlib and Seaborn

Data visualization is a crucial step in the data analysis process. Python offers several libraries for visualization, but the most widely used are **Matplotlib** and **Seaborn**. These libraries help in understanding the data better through graphs, charts, and plots.

---

## 🧰 1. Matplotlib Overview

**Matplotlib** is a low-level plotting library that gives full control over every aspect of a figure. The primary module is `matplotlib.pyplot`.

### Commonly Used Functions in Matplotlib:
- `plot()` - Line plot
- `bar()` - Bar chart
- `scatter()` - Scatter plot
- `hist()` - Histogram
- `pie()` - Pie chart
- `xlabel()`, `ylabel()`, `title()` - For labeling
- `legend()` - To add legends
- `show()` - To display the plot




# 📈 Types of Graphs in Data Visualization and Their Use Cases

Visualizing data is key to understanding trends, spotting outliers, and gaining insights. Below are common types of plots used in **Matplotlib** and **Seaborn**, along with their **use cases**.

---

## 🔹 1. Line Plot

**Function**: `plt.plot()` / `sns.lineplot()`

**Use Case**: Used to show **trends over time** (time-series data) or the relationship between two continuous variables.

## 🔹 2. Bar Plot
**Function**: plt.bar() / sns.barplot()

**Use Case**: Used for comparing categorical data. Ideal for showing group comparisons with height representing value.

✅ Examples:

Number of sales in different regions

Average score by department

## 🔹 3. Count Plot
**Function**: sns.countplot()

**Use Case**: Displays the count of observations in each categorical bin. Useful for categorical frequency analysis.

✅ Examples:

Count of male vs female customers

Number of items in each product category

## 🔹 4. Histogram
**Function**: plt.hist() / sns.histplot()

**Use Case**: Displays the distribution of a single numeric variable by grouping values into bins.

✅ Examples:

Age distribution of users

Distribution of exam scores

## 🔹 5. Box Plot (Box-and-Whisker Plot)
**Function**: sns.boxplot()

**Use Case**: Shows distribution, central tendency, and outliers. Useful for comparing multiple groups.

✅ Examples:

Salary distribution across departments

Performance scores by gender

## 🔹 6. Violin Plot
**Function**: sns.violinplot()

**Use Case**: Combines box plot and KDE (kernel density estimation). Shows distribution shape and outliers.

✅ Examples:

Distribution of income levels by age group

Score distribution by exam type

## 🔹 7. Scatter Plot
**Function**: plt.scatter() / sns.scatterplot()

**Use Case**: Used to show relationships or correlations between two numerical variables.

✅ Examples:

Height vs weight

Advertising spend vs sales

## 🔹 8. Heatmap
**Function**: sns.heatmap()

**Use Case**: Displays a matrix-like structure using colors. Commonly used for correlation matrices or confusion matrices.

✅ Examples:

Correlation between variables

Confusion matrix in ML classification

## 🔹 9. Pie Chart (Only in Matplotlib)
**Function**: plt.pie()

**Use Case**: Used for showing parts of a whole (percentages). Best for a few categories.

✅ Examples:

Market share by brand

Percentage of users by plan

## 🔹 10. Pair Plot
**Function**: sns.pairplot()

**Use Case**: Plots pairwise relationships in a dataset. Ideal for exploring relationships in multivariate data.

✅ Examples:

Iris dataset: comparing sepal and petal measurements

Customer segmentation attributes

## **Summary Table**

| Plot Type    | Use Case                      | Function                     |
| ------------ | ----------------------------- | ---------------------------- |
| Line Plot    | Trends over time              | `plot()`, `lineplot()`       |
| Bar Plot     | Compare categories            | `bar()`, `barplot()`         |
| Count Plot   | Frequency of categories       | `countplot()`                |
| Histogram    | Numeric distribution          | `hist()`, `histplot()`       |
| Box Plot     | Spread & outliers             | `boxplot()`                  |
| Violin Plot  | Distribution shape + outliers | `violinplot()`               |
| Scatter Plot | Correlation between variables | `scatter()`, `scatterplot()` |
| Heatmap      | Matrix visualization          | `heatmap()`                  |
| Pie Chart    | Percentage representation     | `pie()`                      |
| Pair Plot    | Multivariate relationships    | `pairplot()`                 |


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=sns.load_dataset('titanic')
df
```

```python
# Histogram of passenger ages using matplotlib
plt.figure(figsize=(9, 5))
plt.hist(df['age'], bins=40, color='red', edgecolor='black')
plt.title('Distribution of Passenger by Ages')
plt.xlabel('Age of passengers')
plt.ylabel('Count of passengers')
plt.show()
```

```python
sns.scatterplot(x='age',y='fare',data=df,hue='sex')
```

```python
sns.boxplot(y='fare',data=df)
plt.title('Outlier in Age column')
plt.show()
```

```python
plt.plot(df['fare'], marker='+', linestyle='dotted', color='black') 
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
# plt.grid(True)  
plt.show()
```

```python
# Countplot of survival using seaborn
plt.figure(figsize=(6, 4))
sns.countplot(x='survived',data=df)
plt.title('Survival Count (0 = No, 1 = Yes)')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
df.columns
```

```python
sns.barplot(x="class", y="survived", hue="who",data=df,palette='mako')
plt.title("Distribution of passenger class by gender")
```

```python
# Countplot of passenger class using seaborn
plt.figure(figsize=(6, 4))
sns.countplot(x='class', data=df)
plt.title('Passenger Count by Class')
plt.xlabel('Class')
plt.ylabel('Count')
plt.show()
```

```python
# Boxplot of fare by class using seaborn
plt.figure(figsize=(8, 4))
sns.boxplot(x='class', y='fare', data=df)
# sns.boxplot(df)
plt.title('Fare Distribution by Class')
plt.xlabel('Class')
plt.ylabel('Fare')
plt.show()
```

```python
# Scatterplot of age vs fare colored by survival using seaborn
plt.figure(figsize=(8, 4))
sns.scatterplot(x='age', y='fare', hue='survived', data=df)
plt.title('Age vs Fare Colored by Survival')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.show()
```

```python
# Pie chart: Proportion of survivors
plt.figure(figsize=(5, 5))
df['survived'].value_counts().plot.pie(labels=['Not Survived', 'Survived'], autopct='%1.1f%%', colors=["#615de8","#07eff7"])
plt.title('Survival Proportion')
plt.ylabel('')
plt.show()
```

```python

df.corr(numeric_only=True)
```

```python
# Heatmap: Correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".1f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Violin plot: Age distribution by class
plt.figure(figsize=(12, 6))
sns.violinplot(x='class', y='age', data=df)
plt.title('Age Distribution by Class')
plt.xlabel('Class')
plt.ylabel('Age')
plt.show()
```

```python
# Line plot: Average fare by passenger class
plt.figure(figsize=(8, 4))
df.groupby('pclass')['fare'].mean().plot(marker='o')
plt.title('Average Fare by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.xticks([1, 2, 3])
plt.show()
```

```python
# Pair plot: Selected numerical columns
sns.pairplot(df[['age', 'fare', 'pclass', 'survived']].dropna(), hue='survived')
plt.suptitle('Pair Plot of Age, Fare, and Class by Survival', y=1.02)
plt.show()
```
